### 4. Grad-CAM Visual Explainability
Demonstrating model decision visual explanation on a sample test CT image.

```python
sample_path = '../dataset/test/adenocarcinoma/000108.png'
if os.path.exists(sample_path):
    res = generate_gradcam_overlay(sample_path, model_path='../models/best_model.keras')
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    axes[0].imshow(res['original_image'])
    axes[0].set_title('Original CT')
    axes[0].axis('off')
    axes[1].imshow(res['heatmap_colored'])
    axes[1].set_title('Grad-CAM Heatmap')
    axes[1].axis('off')
    axes[2].imshow(res['superimposed_image'])
    axes[2].set_title(f"{res['predicted_class']} ({res['confidence']:.1f}%)")
    axes[2].axis('off')
    plt.tight_layout()
    plt.show()
```
